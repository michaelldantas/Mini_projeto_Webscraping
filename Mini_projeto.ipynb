```python
# Importando as bibliotecas necessárias

from bs4 import BeautifulSoup
import requests
import re
import datetime

# fazendo a requisição e criando o objeto BeautifulSoup
url = "https://coinmarketcap.com/"
requisicao = requests.get(url)
site = BeautifulSoup(requisicao.text,"html.parser")
```

```python
tbody = site.find("tbody")
linhas = tbody.find_all("tr")

# criando um dicionário para armazenar os dados das moedas
moedas = {}

for linha in linhas:
    try:
        nome  = linha.find(class_ = "eWrlhi").text
        codigo = linha.find(class_ = "gOVMPU").text
        valores = linha.find_all(string=re.compile(r"\$"))
        percentuais = linha.find_all(string=re.compile(r"%"))
        preco = valores[0]
        market_cap = valores[2]
        volume = valores[3]

        for i, percentual in enumerate(percentuais):
            if 'dJLZma' in str(percentual.parent):
                percentuais[i] = "-" + str(percentual)
        var_1h = percentuais[0]
        var_24h = percentuais[1]
        var_7d = percentuais[2]
        data = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        dic = {"nome": nome, "codigo": codigo, "preco": preco, 
               "var_1h": var_1h, "var_24h": var_24h, "var_7d": var_7d, 
               "market_cap": market_cap, "volume": volume, "data": data}
        moedas[nome] = dic
      
    except AttributeError:
        break
    except IndexError:
        break
# Dicioário final com os dados das moedas
display(moedas)
```
